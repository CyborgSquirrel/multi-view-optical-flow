```python
import importlib
import logging
import shutil
import subprocess
import sys
from pathlib import Path

IN_COLAB = False
try:
    import google.colab
    IN_COLAB = True
except ModuleNotFoundError:
    pass
print(f"{IN_COLAB=}")

if IN_COLAB:
    logging.basicConfig(level=logging.INFO)
    DRIVE_PATH = Path("/content/drive")
    PROJECT_PATH = DRIVE_PATH / "MyDrive" / "adl4cv"

    def setup():
        from google.colab import drive

        if not DRIVE_PATH.exists():
            google.colab.drive.mount(str(DRIVE_PATH))

        CODE_PATH = Path.cwd() / "code"
        # if CODE_PATH.exists():
        #     shutil.rmtree(CODE_PATH)
        # if not (DRIVE_PATH / "code.zip").exists():
        shutil.copy(PROJECT_PATH / "code.zip", ".")
        subprocess.run(["unzip", "-o", "-q", "code.zip"], check=True)
        !cat code/id.txt

        # Reload modules
        for module in list(sys.modules.values()):
            if (
                   not hasattr(module, "__file__")
                or module.__file__ is None
                or not module.__file__.startswith(f"{CODE_PATH}/")
            ):
                continue
            logging.info("Reloading %r...", module)
            importlib.reload(module)

        if str(CODE_PATH) not in sys.path:
            sys.path.append(str(CODE_PATH))

    setup()
```
